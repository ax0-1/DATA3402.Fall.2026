````python
def computer_move(game, player):
    n = len(game)
    opponent = 2 if player == 1 else 1

    # win
    for row in range(n):
        for col in range(n):
            if game[row][col] == 0:
                game[row][col] = player
                if winner(game) == player:
                    return
                game[row][col] = 0

    # block
    for row in range(n):
        for col in range(n):
            if game[row][col] == 0:
                game[row][col] = opponent
                if winner(game) == opponent:
                    game[row][col] = player
                    return
                game[row][col] = 0

    # claude said this section is needed for "fallback"
    for row in range(n):
        for col in range(n):
            if game[row][col] == 0:
                game[row][col] = player
                return

game = [[1, 1, 0], [2, 2, 0], [0, 0, 0]]
computer_move(game, 1)
print(game)
````
