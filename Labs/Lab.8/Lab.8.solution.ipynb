### Exercise 3: Training a Classifier

Let's repeat what we did by hand in the previous lab (Lab 7, Exercise 8) using scikit-learn. We'll use an LDA classifier, which we can instantiate as follows:

```python
import sklearn.discriminant_analysis as DA
Fisher=DA.LinearDiscriminantAnalysis()
```

As discussed in the lecture, to properly formulate our problem, we'll have to:

* Define the inputs (X) vs outputs (Y)
* Designate training vs testing samples (in order to get an unbiased assessment of the performance of machine learning algorithms)

For example, here we'll use 4M events for training and the remainder for testing.

```python
N_Train=4000000

Train_Sample=df[:N_Train]
Test_Sample=df[N_Train:]

X_Train=Train_Sample[VarNames[1:]]
y_Train=Train_Sample["signal"]

X_Test=Test_Sample[VarNames[1:]]
y_Test=Test_Sample["signal"]

Test_sig=Test_Sample[Test_Sample.signal==1]
Test_bkg=Test_Sample[Test_Sample.signal==0]
```

We can train the classifier as follows:

```python
Fisher.fit(X_Train,y_Train)
```

We can plot the output, comparing signal and background:

```python
plt.figure()
plt.hist(Fisher.decision_function(Test_sig[VarNames[1:]]),bins=100,histtype="step", color="blue", label="signal",stacked=True)
plt.hist(Fisher.decision_function(Test_bkg[VarNames[1:]]),bins=100,histtype="step", color="red", label="background",stacked=True)
plt.legend(loc='upper right')
plt.show()
```

#### Part a

Compare the ROC curves computed on the test and on the training samples, in a single plot. Do you see a bias? (Use `Fisher.decision_function` on each sample, and `roc_curve` as in Exercise 5.)

```python
from sklearn.metrics import roc_curve, auc

fpr_train, tpr_train, _ = roc_curve(
    y_Train, Fisher.decision_function(X_Train))

fpr_test, tpr_test, _ = roc_curve(
    y_Test, Fisher.decision_function(X_Test))

plt.plot(fpr_train, tpr_train, label="Training")
plt.plot(fpr_test, tpr_test, label="Testing")

plt.xlabel("False Positive Rate")
plt.ylabel("True Positive Rate")
plt.legend()
plt.show()
```

#### Part b

Train three Fisher discriminants, using as input (1) only the raw quantities (`RawNames`), (2) only the derived features (`FeatureNames`), and (3) all 18. Compare their ROC curves, computed on the test sample, in a single plot. Which inputs carry the separation?

```python
from sklearn.metrics import roc_curve

Fisher_Raw = DA.LinearDiscriminantAnalysis()
Fisher_Features = DA.LinearDiscriminantAnalysis()

Fisher_Raw.fit(Train_Sample[RawNames], y_Train)
Fisher_Features.fit(Train_Sample[FeatureNames], y_Train)

fpr_raw, tpr_raw, _ = roc_curve(y_Test, Fisher_Raw.decision_function(Test_Sample[RawNames]))

fpr_features, tpr_features, _ = roc_curve(y_Test, Fisher_Features.decision_function(Test_Sample[FeatureNames]))

fpr_all, tpr_all, _ = roc_curve(y_Test, Fisher.decision_function(X_Test))

plt.figure()
plt.plot(fpr_raw, tpr_raw, label="Raw Quantities")
plt.plot(fpr_features, tpr_features, label="Derived Features")
plt.plot(fpr_all, tpr_all, label="All 18")
plt.xlabel("False Positive Rate")
plt.ylabel("True Positive Rate")
plt.legend()
plt.show()
```

#### Part c

Use the best method from part b to compute the maximal significance $\sigma_S= \frac{N_S}{\sqrt{N_S+N_B}}$ for the four scenarios in Lab 7 (Exercise 5.3), choosing the cut on the classifier's output that maximizes it.

```python
from sklearn.naive_bayes import GaussianNB
from sklearn.tree import DecisionTreeClassifier
from sklearn.linear_model import SGDClassifier
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler

classifiers = {
    "GaussianNB": GaussianNB(),
    "Decision Tree": DecisionTreeClassifier(max_depth=8, random_state=42),
    "SGD": make_pipeline(StandardScaler(), SGDClassifier(loss="log_loss", max_iter=1000, tol=1e-3, random_state=42))
}
```

```python
from sklearn.base import clone
from sklearn.metrics import roc_curve, auc

datasets = {
    "Raw Quantities": (Train_Sample[RawNames], Test_Sample[RawNames]),
    "Derived Features": (Train_Sample[FeatureNames], Test_Sample[FeatureNames]),
    "All 18": (X_Train, X_Test)
}

def compare_classifier(classifier):
    output = {}
    plt.figure(figsize=(8,6))

    for name, (x_train, x_test) in datasets.items():
        model = clone(classifier)
        model.fit(x_train, y_Train)

        if hasattr(model, "decision_function"):
            scores = model.decision_function(x_test)
        else:
            scores = model.predict_proba(x_test)[:, 1]

        fpr, tpr, _ = roc_curve(y_Test, scores)
        roc_auc = auc(fpr, tpr)

        plt.plot(fpr, tpr, label=f"{name} (AUC={roc_auc:.3f})")
        output[name] = {"model": model, "auc": roc_auc}

    plt.xlabel("False Positive Rate")
    plt.ylabel("True Positive Rate")
    plt.legend()
    plt.show()

    return output

results = {}

for name, classifier in classifiers.items():
    results[name] = compare_classifier(classifier)
```

```python
best_classifier, best_inputs = max(
    [(c, d) for c, comparison in results.items() for d in comparison],
    key=lambda x: results[x[0]][x[1]]["auc"]
)

best_model = results[best_classifier][best_inputs]["model"]
best_test = datasets[best_inputs][1]

if hasattr(best_model, "decision_function"):
    scores = best_model.decision_function(best_test)
else:
    scores = best_model.predict_proba(best_test)[:, list(best_model.classes_).index(1)]

fpr, tpr, thresholds = roc_curve(y_Test, scores)

scenarios = [(10, 100), (100, 1000), (1000, 10000), (10000, 100000)]

print(best_classifier, best_inputs)

for NS, NB in scenarios:
    ns = tpr * NS
    nb = fpr * NB
    significance = np.divide(ns, np.sqrt(ns + nb), out=np.zeros_like(ns), where=(ns + nb) > 0)
    i = np.argmax(significance)
    print(NS, NB, significance[i], thresholds[i])
```

### Exercise 5: Metrics

Scikit-learn provides methods for computing the FPR, TPR, ROC, AUC metrics. For example:

#### Part b
Look through [model evaluation](https://scikit-learn.org/stable/modules/model_evaluation.html#) documentation. Using scikit-learn's tools, compute TPR, FPR, ROC, AUC, Precision, Recall, F1 score, and accuracy for the method you selected in Exercise 4c and each scenario. Make a nice table, which also includes the maximal significance. 


```python
from sklearn.metrics import roc_curve, auc, precision_score, recall_score, f1_score, accuracy_score, confusion_matrix

scores = best_model.decision_function(best_test)

fpr, tpr, thresholds = roc_curve(y_Test, scores)
roc_auc = auc(fpr, tpr)

scenarios = [(10, 100), (100, 1000), (1000, 10000), (10000, 100000)]

rows = []

for NS, NB in scenarios:
    ns = tpr * NS
    nb = fpr * NB
    significance = np.divide(ns, np.sqrt(ns + nb), out=np.zeros_like(ns), where=(ns + nb) > 0)
    i = np.argmax(significance)

    y_pred = (scores >= thresholds[i]).astype(int)

    tn, fp, fn, tp = confusion_matrix(y_Test, y_pred, labels=[0, 1]).ravel()

    rows.append({
        "Scenario": f"{NS}/{NB}",
        "TPR": recall_score(y_Test, y_pred),
        "FPR": fp / (fp + tn),
        "AUC": roc_auc,
        "Precision": precision_score(y_Test, y_pred, zero_division=0),
        "Recall": recall_score(y_Test, y_pred),
        "F1": f1_score(y_Test, y_pred),
        "Accuracy": accuracy_score(y_Test, y_pred),
        "Max Significance": significance[i]
    })

plt.plot(fpr, tpr, label=f"SGD All 18 (AUC={roc_auc:.3f})")
plt.xlabel("False Positive Rate")
plt.ylabel("True Positive Rate")
plt.legend()
plt.show()

display(pd.DataFrame(rows).round(4))
```
